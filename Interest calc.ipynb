```python
loan  = 1000000
intrest = 0.1
result1 = input('what method would you like \n 1)Monthly \n 2) Weekly \n 3) Daily ')
result2 = int(input(f'Your {result1} payment is: '))
if result1 == 'Monthly':
    p = 12
if result2 == 'Weekly':
    p = 52
else :
    p =365
p_int = intrest/p
```

```python
time = 0
while loan>0:
    loan = loan-result2
    loan = loan + loan*p_int
    time += 1
print(f'You took {time} {result1} to pay your loan')

```
